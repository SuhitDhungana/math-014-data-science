## Part B: Data Wrangling

### 6. Add a New Column Using Existing Columns

```python
df_new = df.copy()
```

```python
low_unemployment_rank = df_new["unemployment_rate"].quantile(1/3)
medium_unemployment_rank = df_new["unemployment_rate"].quantile(2/3)


def classify_unemployment(unemployment_rate):
    if unemployment_rate < low_unemployment_rank:
        return "Low"
    elif unemployment_rate < medium_unemployment_rank:
        return "Medium"
    else:
        return "High"


df_new["unemployment_rank"] = df_new["unemployment_rate"].apply(classify_unemployment)
df_new["unemployment_rank"].head()
```

This new column `unemployment_rank` classifies the dataset into groups based on the unemployment rate. States with unemployment rate less than 1/3 quantile are classified as "Low," between 1/3 and 2/3 quantiles are classified as "Medium," and the ones who have unemployment rate are classified above 2/3 quantile are classified as "High."

### 7. Filter the Data

```python
# A. Filtering out the crimes that occured in the DMV Area
df_filt_DMV = df_new[df_new["state"].isin(["maryland", "virginia", "district of columbia"])]
df_filt_DMV = df_filt_DMV[["state_abb", "year", "crime_category", "count", "reporting_population", "rate_per_100k_reporting_pop", "unemployment_rank"]]
df_filt_DMV.head()
```

- #### Use of sub-dataset

The sub-dataset `df_filt_DMV` consists of the crimes that have happened in the DMV area. This subset excludes other states and allows for exclusive study of only the DMV area. It can be also be used to compare the crimes that occur in the DMV area to the crimes that occur in other states.

```python
# B. Filtering out the crimes that occured during the COVID-19 
df_filt_covid = df_new[(df_new["year"] >= 2020) & (df_new["year"] <= 2023)]
df_filt_covid = df_filt_covid[["state_abb", "year", "crime_category", "count", "reporting_population", "rate_per_100k_reporting_pop", "unemployment_rank"]]
df_filt_covid.head()
```

- #### Use of sub-dataset

The sub-dataset `df_filt_covid` consists of the crimes that have been committed during the Covid-19 period. This subset is useful to study the change of the crime rates during the Covid-19 period as compared to other times. Covid-19 period may also be a potential outlier, as the US economy was in recession and unemployment rates were at an all-time high.

```python
# C. Filtering out the crimes recorded in state-years with high poverty (poverty rate > 18)
df_filt_low_income = df_new[(df_new["acs1_poverty_rate"] > 18)]
df_filt_low_income = df_filt_low_income[["state_abb", "year", "crime_category", "count", "reporting_population", "rate_per_100k_reporting_pop", "unemployment_rank", "acs1_poverty_rate", "acs1_median_household_income"]]
df_filt_low_income.head()
```

- #### Use of sub-dataset

The sub-dataset `df_filt_low_income` consists of the crimes that have been in state-years where there was high poverty (poverty rate > 18%). This subset is going to be useful to find out how much poverty can affect one's tendency to commit crimes. This subset can also be used to find the correlation of high poverty, unemployment, and the crime rates as well.

### 9. Grouping and Aggregation

- #### Grouping data by one important categorical column

```python
# Grouping the low-income dataset according to the crime category
crimes_unemployment_total = df_filt_low_income.groupby(["crime_category", "unemployment_rank"])["count"].sum()
print("The total number of crimes of different categories by people of different unemployment rank of the state-months are mentioned below:\n", crimes_unemployment_total)
print()

# Finding the total sums of crimes for a particular crime category (doesn't take into account the unemployment rank)
crimes_totals = df_filt_low_income.groupby("crime_category")["count"].sum()
print("The total number of crimes for each category is shown below: ", crimes_totals)
```

- #### Finding one summary measure (percentage of crimes for each rank within each category)

```python
# Finding the percentage of crimes for each rank within each category
rows = []
for (crime, unemp_rank), crime_group_total in crimes_unemployment_total.items():
    crime_unemployment_percentage = crime_group_total / crimes_totals[crime] * 100
    rows.append([crime, unemp_rank, crime_group_total, round(crime_unemployment_percentage, 2)])
rows
```

- #### Display the grouped result

```python
crimes_percentage_df = pd.DataFrame(rows, columns=["crime_category", "unemployment_rank", "total_crimes", "crime_percentage"])
print(crimes_percentage_df[["crime_category", "unemployment_rank", "crime_percentage"]])
```

- #####  Observation

Among people in high poverty subset, larceny theft is the most reported crime (7304989 instances), whereas murder is the least reported crime (24164 instances). Additionally, at least 66.65% of the crimes of every crime type occurred in state-months with "High" unemployment.

- #### Another grouped data (related to Covid 19 dataset)

```python
crime_rates_by_year = df_filt_covid.groupby("year")["rate_per_100k_reporting_pop"].agg(["mean", "min", "max", "count"])
crime_rates_by_year
```

During Covid-19, crime rates fluctuated from 26.2 to 29.6 crimes per 100,000 people per month, where it was highest in 2020 and lowest in 2021.
